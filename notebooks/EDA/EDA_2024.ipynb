# Carregamento e Seleção

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Carregar o arquivo 
planilhas = pd.read_excel('data/base_dados_2024.xlsx', sheet_name=None)

# Selecionar os dados de 2024
df_24 = planilhas['PEDE2024']
```

# Diagnóstico de Qualidade

```python
# Verificação de tipos e nulos
print(df_24.info())

# Cálculo de porcentagem de nulos
nulos_qtd = df_24.isnull().sum()
nulos_perc = (nulos_qtd / len(df_24)) * 100
df_nulos_24 = pd.DataFrame({'Qtd_Nulos': nulos_qtd, 'Porcentagem (%)': nulos_perc})
print(df_nulos_24[df_nulos_24['Qtd_Nulos'] >= 0].sort_values(by='Qtd_Nulos', ascending=False))
```

```python
# Contagem de quem entrou até 2022 
ingressantes_ate_2022 = df_24[df_24['Ano ingresso'] <= 2023].shape[0]

# Contagem de quem entrou em 2023 
ingressantes_2023 = df_24[df_24['Ano ingresso'] >= 2024].shape[0]

print(f"Alunos que entraram até 2022: {ingressantes_ate_2022}")
print(f"Alunos que entraram a partir de 2023: {ingressantes_2023}")
print()
print(f"Total de alunos na base: {len(df_24)}")

# Para ver o detalhamento ano a ano:
print("\nDistribuição detalhada por ano de ingresso:")
print(df_24['Ano ingresso'].value_counts().sort_index())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd


# Em 2024, o INDE e as Notas costumam vir como object (texto) devido a vírgulas
cols_numericas = ['Idade', 'INDE 2024', 'Mat', 'Por', 'Ing', 'IPV']
for col in cols_numericas:
    if col in df_24.columns:
        df_24[col] = pd.to_numeric(df_24[col], errors='coerce')


def formatar_fase(x):
    if pd.isna(x): return x
    x_str = str(x).strip()
    return f"Fase {x_str}" if x_str.isdigit() else x_str

df_24['Fase_Plot'] = df_24['Fase'].apply(formatar_fase)

# =========================================================
# 1. Distribuição de Gênero
# =========================================================
plt.figure(figsize=(8, 6))
genero_counts = df_24['Gênero'].value_counts()
plt.pie(genero_counts, labels=genero_counts.index, autopct='%1.1f%%', startangle=90, colors=sns.color_palette('pastel'))
plt.title("Distribuição de Gênero (2024)", fontsize=14)
plt.show()

# =========================================================
# 2. Distribuição de Idade
# =========================================================
plt.figure(figsize=(10, 6))
sns.histplot(df_24['Idade'].dropna(), bins=15, kde=True, color='skyblue')
plt.title("Distribuição de Idade dos Alunos (2024)", fontsize=14)
plt.xlabel("Idade (Anos)", fontsize=12)
plt.ylabel("Frequência", fontsize=12)
plt.show()

# =========================================================
# 3. Média do INDE por Fase
# =========================================================
order_fase = ['ALFA', 'Fase 1', 'Fase 2', 'Fase 3', 'Fase 4', 'Fase 5', 'Fase 6', 'Fase 7', 'Fase 8']

plt.figure(figsize=(12, 6))
sns.barplot(
    x='Fase_Plot', 
    y='INDE 2024', 
    data=df_24, 
    order=order_fase, 
    palette='viridis', 
    estimator=np.mean, 
    errorbar=None
)

plt.title("Média do INDE por Fase (2024)", fontsize=16, fontweight='bold')
plt.xlabel("Fase (Nível PM)", fontsize=12)
plt.ylabel("Média do INDE", fontsize=12)
plt.xticks(rotation=45)

ax = plt.gca()
for p in ax.patches:
    if p.get_height() > 0:
        ax.annotate(f'{p.get_height():.2f}', 
                    (p.get_x() + p.get_width() / 2., p.get_height()), 
                    ha = 'center', va = 'center', 
                    xytext = (0, 9), 
                    textcoords = 'offset points',
                    fontsize=10, fontweight='bold')

plt.tight_layout()
plt.show()

# =========================================================
# 4. Classificação por Pedra
# =========================================================
order_pedra = ['Quartzo', 'Ágata', 'Ametista', 'Topázio']

plt.figure(figsize=(10, 6))
sns.countplot(x='Pedra 2024', data=df_24, order=order_pedra, palette='coolwarm')
plt.title("Classificação dos Alunos por Pedra (2024)", fontsize=14)
plt.xlabel("Pedra 2024", fontsize=12)
plt.ylabel("Contagem", fontsize=12)
plt.show()

# =========================================================
# 5. Comparativo Matemática vs Português
# =========================================================
notas_core = df_24[['Mat', 'Por']].mean().reset_index()
notas_core.columns = ['Matéria', 'Média']
notas_core['Matéria'] = notas_core['Matéria'].replace({'Mat': 'Matemática', 'Por': 'Português'})

plt.figure(figsize=(8, 6))
sns.barplot(x='Matéria', y='Média', data=notas_core, palette='Set2', errorbar=None)

plt.title("Comparativo de Médias: Matemática vs Português (2024)", fontsize=16)
plt.ylabel("Média da Nota (0-10)", fontsize=12)
plt.xlabel("Disciplina", fontsize=12)
plt.ylim(0, 10) 

for i, v in enumerate(notas_core['Média']):
    plt.text(i, v + 0.2, f"{v:.2f}", ha='center', fontweight='bold')

plt.show()

# =========================================================
# 6. Desempenho: Gênero vs INDE
# =========================================================
plt.figure(figsize=(10, 6))
sns.boxplot(x='Gênero', y='INDE 2024', data=df_24, palette='pastel')

plt.title("Comparativo de Desempenho: Gênero vs INDE (2024)", fontsize=16)
plt.ylabel("INDE 2024", fontsize=12)
plt.show()

# =========================================================
# 7. Proporção do Ponto de Virada
# =========================================================
# Criando a coluna temporária para 2024 usando o IPV
df_24['Temp_PV_24'] = df_24['IPV'].apply(lambda x: 'Sim' if pd.notnull(x) and x >= 8.0 else ('Não' if pd.notnull(x) else np.nan))

plt.figure(figsize=(8, 6))
pv_counts = df_24['Temp_PV_24'].dropna().value_counts()

if not pv_counts.empty:
    plt.pie(pv_counts, labels=pv_counts.index, autopct='%1.1f%%', startangle=140, colors=['#ff9999','#66b3ff'])
    plt.title("Proporção de Alunos que Atingiram o 'Ponto de Virada' (2024)", fontsize=14)
    plt.show()

# =========================================================
# 8. Relação Idade vs Performance
# =========================================================
plt.figure(figsize=(12, 7))
sns.scatterplot(x='Idade', y='INDE 2024', hue='Pedra 2024', data=df_24, palette='viridis', s=100, alpha=0.7)

plt.title("Relação Idade vs Performance (Colorido por Pedra) - 2024", fontsize=16)
plt.xlabel("Idade em 2024", fontsize=12)
plt.ylabel("Nota INDE 2024", fontsize=12)
plt.legend(title='Classificação', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

print("="*50)
print("🔍 INVESTIGAÇÃO: O MISTÉRIO DA PEDRA ÁGATA (EDIÇÃO 2024)")
print("="*50)

# ---------------------------------------------------------
# PREPARAÇÃO: Conversão do INDE 2024 para numérico
# ---------------------------------------------------------
# Necessário porque em 2024 o dado pode vir com vírgulas ou como objeto
df_24['INDE 2024'] = pd.to_numeric(df_24['INDE 2024'], errors='coerce')

# ---------------------------------------------------------
# ANÁLISE: Tabela de limites de notas (INDE) por Pedra em 2024
# ---------------------------------------------------------
# Isso vai mostrar se os intervalos de nota para Ágata mudaram em 2024
limites_pedra_24 = df_24.groupby('Pedra 2024', observed=False)['INDE 2024'].agg(['min', 'max', 'count'])
print("\n--- Faixas de Nota (INDE) por Pedra em 2024 ---")
print(limites_pedra_24.sort_values(by='min'))
print("\n" + "="*50)

# ---------------------------------------------------------
# VISUALIZAÇÃO: Histograma detalhado do INDE 2024
# ---------------------------------------------------------
plt.figure(figsize=(12, 6))

# Plotando a distribuição das notas de 2024
sns.histplot(df_24['INDE 2024'].dropna(), bins=40, kde=True, color='darkorange', edgecolor='black')

plt.title("Distribuição das Notas INDE (2024) - Investigando o 'Buraco' da Ágata", fontsize=15, fontweight='bold')
plt.xlabel("Nota INDE 2024 (0 a 10)", fontsize=12)
plt.ylabel("Quantidade de Alunos", fontsize=12)

# Adicionando marcações visuais nos eixos
plt.xticks(range(0, 11))
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```
