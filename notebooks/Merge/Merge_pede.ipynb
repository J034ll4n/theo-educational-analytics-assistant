# Aqui vamos fazer um Merge do dataset

```python
import pandas as pd

# Carregar os arquivos finais que você tratou
df22 = pd.read_csv('data/pede_2022_final.csv')
df23 = pd.read_csv('data/pede_2023_final.csv')
df24 = pd.read_csv('data/pede_2024_final.csv')

# Resolver o Status para 2022 e 2023 (1 = Ativo)
# Como são bases de ciclos encerrados, todos nela são considerados ativos no período
df22['status_aluno'] = 1
df23['status_aluno'] = 1


# MERGE (CONCATENAR) - Empilhando os anos
# O ignore_index=True cria um novo índice sequencial para a base mestre
df_master = pd.concat([df22, df23, df24], ignore_index=True)

# 6. Salvar o arquivo final unificado
df_master.to_csv('pede_master.csv', index=False, encoding='utf-8-sig')

print(f" Status resolvido e bases unificadas!")
print(f" Total de registros na base Master: {df_master.shape[0]}")
print(f" Anos presentes: {df_master['ano_referencia'].unique()}")
```
