# Carregamento e Padronização de Textos (Strings)

```python
import pandas as pd
import numpy as np

df = pd.read_csv('main.csv')
text_columns = [
    'fase', 'turma', 'genero', 'instituicao_de_ensino', 
    'pedra_2020', 'pedra_2021', 'pedra', 'fase_ideal', 
    'serie_escolar', 'pedra_2022', 'atingiu_pv', 
    'indicado_bolsa', 'pedra_2023', 'escola'
]

for col in text_columns:
    if col in df.columns:
        df[col] = df[col].astype(str).str.strip().str.title().replace('Nan', np.nan)

print("Textos padronizados com sucesso!")
```

# Tratamento de Datas e Idade

```python


# Padronizar a coluna de data de nascimento para o formato datetime do Pandas
df['data_de_nasc'] = pd.to_datetime(df['data_de_nasc'], format='mixed', errors='coerce')

# Preencher os anos de nascimento vazios extraindo o ano da data_de_nasc
df['ano_nasc'] = df['ano_nasc'].fillna(df['data_de_nasc'].dt.year)

#  Limpar a coluna 'idade'
df['idade'] = pd.to_numeric(df['idade'].astype(str).str.extract(r'(\d+)')[0], errors='coerce')

#  Preencher anos de nascimento vazios para quem só tinha a idade informada
df['ano_nasc'] = df['ano_nasc'].fillna(df['ano_referencia'] - df['idade'])

#  Preencher idades vazias (ano de referência - ano_nasc)
df['idade'] = df['idade'].fillna(df['ano_referencia'] - df['ano_nasc'])

#  RESOLVER O NaT (Datas Vazias): Criar uma data genérica (1º de Julho do ano) para quem não tem data_de_nasc
datas_imputadas = pd.to_datetime(df['ano_nasc'].astype(int).astype(str) + '-07-01', errors='coerce')
df['data_de_nasc'] = df['data_de_nasc'].fillna(datas_imputadas)

#  Remover a hora da data de nascimento, mantendo apenas Ano-Mês-Dia (YYYY-MM-DD)
df['data_de_nasc'] = df['data_de_nasc'].dt.date

#  Converter o ano de nascimento para número inteiro
df['ano_nasc'] = df['ano_nasc'].astype(int)

print("Datas, idades, anos de nascimento e NaTs tratados com sucesso!")
```

# Garantia de Tipagem das Variáveis Numéricas (Indicadores)

```python
#  Lista de colunas numéricas (indicadores e notas)
numerical_cols = [
    'inde', 'cg', 'cf', 'ct', 'iaa', 'ieg', 
    'ips', 'ida', 'mat', 'por', 'ing', 'ipv', 
    'ian', 'ipp'
]

#  Forçar a conversão para float, transformando qualquer erro/texto acidental em NaN
for col in numerical_cols:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors='coerce')


# Verificar como ficou a estrutura final
print("\n--- INFO DO DATASET PADRONIZADO ---")
print(df.info())
```

# Correção de valores nulos de alunos que entraram recentemente

```python
import pandas as pd

# Preenchendo os nulos dos anos anteriores (TEXTO) com uma categoria específica
colunas_pedra_antigas = ['pedra_2020', 'pedra_2021', 'pedra_2022', 'pedra_2023']
for col in colunas_pedra_antigas:
    df[col] = df[col].fillna('Sem_Historico')

# Correção de nulos numéricos antigos (NÚMERO) - Usando -999 em vez de texto como uma estratégia para posteriormente o modelo perceber que está totalmente fora do padrão normal. 
df['inde_2022'] = df['inde_2022'].fillna(-999)
df['inde_2023'] = df['inde_2023'].fillna(-999)

# Removendo linhas onde falta o 'gabarito' (pedra/inde) e a instituicao (1 linha)
df = df.dropna(subset=['pedra', 'inde', 'instituicao_de_ensino'])

# Preenchendo categorias de texto com 'Desconhecido'
colunas_texto = ['escola', 'indicado_bolsa', 'atingiu_pv']
for col in colunas_texto:
    df[col] = df[col].fillna('Desconhecido')

#  Preenchendo as notas restantes (como o ipp e o que sobrou) com a Mediana
colunas_numericas = ['ipp', 'ieg', 'iaa', 'ips', 'ida', 'mat', 'por', 'ipv']
for col in colunas_numericas:
    # A mediana ignora os nulos para calcular o valor do meio, depois preenche os vazios
    df[col] = df[col].fillna(df[col].median())

# Verificando o resultado
print("Total de nulos restantes na base:", df.isna().sum().sum())
```

# Padronização

```python
# Padronização Nominal
mapeamento_escolas = {
    'Faculdade Fiap': 'FIAP',
    'Fiap School - Sp': 'FIAP',
    'Ffiap': 'FIAP',
    'EE Chácara Florida II': 'EE Chácara Florida'
}
df['escola'] = df['escola'].replace(mapeamento_escolas)


#Tipagem de Colunas Booleanas 

mapa_bool = {'Sim': True, 'Não': False, 1: True, 0: False, '1': True, '0': False}
colunas_bool = ['tem_ingles', 'atingiu_pv', 'indicado_bolsa']

for col in colunas_bool:
    df[col] = df[col].map(mapa_bool).astype(bool)
```

```python

#  Excluir as colunas ano_nasc e data_de_nasc
colunas_para_remover = ['ano_nasc', 'data_de_nasc, data_nasc']
df = df.drop(columns=[col for col in colunas_para_remover if col in df.columns])

# Renomear a coluna data_nasc_limpa para data_nasc
if 'data_nasc_limpa' in df.columns:
    df = df.rename(columns={'data_nasc_limpa': 'data_nasc'})
```
